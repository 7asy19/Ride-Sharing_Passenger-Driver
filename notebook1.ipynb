```python
import csv
import random
import math
import matplotlib.pyplot as plt

def distance(passenger, driver):
    return math.sqrt((passenger[1] - driver[1]) ** 2 + (passenger[2] - driver[2]) ** 2)

drivers = []
passengers = []

file_path = r'D:\intro ai final project\Ride_sharing_Passenger_Driver.csv'

with open(file_path, mode="r", encoding="utf-8") as file:
    data = csv.DictReader(file)
    for row in data:
        if row["TYPE"] == "Driver":
            drivers.append((row["ID"], int(row["X"]), int(row["Y"])))
        else:
            if row["TYPE"] == "Passenger":
                passengers.append((row["ID"], int(row["X"]), int(row["Y"])))

matched_pairs = []
matches_for_plot = []
unmatched_passengers = passengers.copy()
for driver in drivers:
    closest_passenger = min(unmatched_passengers, key=lambda passenger : distance(passenger, driver))
    dist = distance(closest_passenger, driver)
    matched_pairs.append((driver[0], closest_passenger[0], dist))
    matches_for_plot.append({
        'driver': driver,
        'passenger': closest_passenger
    })
    unmatched_passengers.remove(closest_passenger)
total_distance = sum(pair[2] for pair in matched_pairs)
average_distance = total_distance / len(matched_pairs)
for match in matched_pairs:
    print(match, end="\n")
print("Total pickup distance:", total_distance)
print("Average pickup distance:", average_distance)
print("Unmatched passengers:", [p[0] for p in unmatched_passengers])

random_matches = []
total_random_dist=0
random.seed(42)
shuffled_passengers = passengers.copy()
random.shuffle(shuffled_passengers)
for i in range(len(drivers)):
    d=drivers[i]
    p=shuffled_passengers[i]
    dist=distance(p,d)
    random_matches.append((d[0],p[0],dist))
    total_random_dist += dist

for random_match in random_matches:
    print(random_match,end="\n")

print(f"random matching distance:{total_random_dist}" )

###############################################################################
plt.figure(figsize=(15, 10))
ax1 =plt.subplot2grid((2,2),(0,0))
ax2 =plt.subplot2grid((2,2),(1,0))
ax3 =plt.subplot2grid((2,2),(0,1), rowspan = 2)

pass_x = [p[1] for p in passengers]
pass_y = [p[2] for p in passengers]
ax1.scatter(pass_x, pass_y, color='mediumblue', label='Passengers', alpha=1.0, s=50 , edgecolors = 'black')


for p in passengers:
    ax1.text(p[1] + 1, p[2] + 1, p[0], color="mediumblue", fontsize=9 , fontweight = 'bold', alpha=1.0)


driver_x = [d[1] for d in drivers]
driver_y = [d[2] for d in drivers]
ax1.scatter(driver_x, driver_y,  color='Darkgray', marker='d', label='Drivers', s=80, alpha=1.0,edgecolors = 'black')


for d in drivers:
    ax1.text(d[1] - 3, d[2] - 3, d[0], color='black', fontweight='bold', fontsize=9, alpha=1.0)

for match in matches_for_plot:
    d = match['driver']
    p = match['passenger']

    ax1.plot([d[1], p[1]], [d[2], p[2]], '--', alpha=0.7, linewidth=1.5 , color = 'black')


ax1.set_title(' Greedy Matching Assignments', fontsize=14)
ax1.legend(loc='upper right')
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.set_xlim(-5, 105)
ax1.set_ylim(-5, 105)
  #################################################
ax2.scatter([p[1] for p in passengers], [p[2] for p in passengers], color='mediumblue', alpha=1.0, s=50 , edgecolors = 'black' , label='Passengers')
for p in passengers:
    ax2.text(p[1] + 1, p[2] + 1, p[0], color='mediumblue', fontsize=8 , fontweight = 'bold')

ax2.scatter([d[1] for d in drivers], [d[2] for d in drivers], color='darkgray', marker='d', s=80 , edgecolors = 'black' ,label = 'Drivers' )
for d in drivers:
    ax2.text(d[1] - 3, d[2] - 3, d[0], color='black', fontweight='bold', fontsize=9)

for match in random_matches:
    d = match[0]
    p = match[1]
    ax2.plot([d[1], p[1]], [d[2], p[2]], '--', alpha=0.7, linewidth=1.5 , color = 'black')
ax2.set_title('Random Matching Assignments', fontsize=14)
ax2.legend(loc='upper right')
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.set_xlim(-5, 105)
ax2.set_ylim(-5, 105)

######################################################

labels = ['Greedy Distance', 'Random Distance']
totals = [total_distance, total_random_dist]
average_greedy = total_distance / len(drivers)
average_random = total_random_dist / len(drivers)
ax3.bar(labels, totals, color=['mediumblue', 'mediumblue'], alpha=1.0)
ax3.set_title('Total & Average Distance')

ax3.text(0, total_distance + 1, f'{total_distance:.2f}', ha='center', fontweight='bold')
ax3.text(1, total_random_dist + 1, f'{total_random_dist:.2f}', ha='center', fontweight='bold')
ax3.text(0, total_distance/2, f'Avg: {average_greedy:.2f}', ha='center', color='darkgray', fontweight='bold')
ax3.text(1, total_random_dist/2, f'Avg: {average_random:.2f}', ha='center', color='darkgray', fontweight='bold')
plt.tight_layout()
plt.show()
```
